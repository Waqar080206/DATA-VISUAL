```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
df = pd.read_csv("boston_house_prices.csv", header=1)

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="RM",
    y="MEDV"
)

plt.title("Average Number of Rooms vs House Value")
plt.xlabel("Average Number of Rooms (RM)")
plt.ylabel("Median House Value (MEDV)")
plt.grid(alpha=0.3)

plt.show()
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="NOX",
    y="MEDV"
)

plt.title("Pollution vs  Value")
plt.xlabel("Nitric Oxide Concentration (NOX)")
plt.ylabel(" Value (MEDV)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="PTRATIO",
    y="MEDV"
)

plt.title("Pupil-Teacher Ratio vs House Value")
plt.xlabel(" (PTRATIO)")
plt.ylabel("Value (MEDV)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="LSTAT",
    y="MEDV"
)

plt.title("LSTAT vs House Value")

plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="RM",
    y="LSTAT"
)

plt.title("Number of Rooms vs LSTAT")
plt.xlabel("Average Number of Rooms (RM)")
plt.ylabel("Lower Status Population (%) - LSTAT")
plt.grid(alpha=0.3)

plt.show()
```

```python
correlation_matrix = df.corr(numeric_only=True)

print(correlation_matrix.round(2))
```

```python
plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Boston Housing Dataset")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="RAD",
    y="TAX"
)

plt.title("RAD vs TAX")
plt.xlabel("Accessibility to Radial Highways (RAD)")
plt.ylabel("Property Tax Rate (TAX)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="LSTAT",
    y="MEDV"
)

plt.title("LSTAT vs MEDV")
plt.xlabel("Lower Status Population (%) - LSTAT")
plt.ylabel("Median House Value (MEDV)")
plt.grid(alpha=0.3)

plt.show()
```
