```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

iris = load_iris()
X = iris.data
y = iris.target

df1 = pd.DataFrame(X, columns=iris.feature_names)
df1["Species"] = iris.target_names[y]

sns.pairplot(df1, hue="Species")
plt.show()
```

```python
pca3 = PCA(n_components=3)
X3 = pca3.fit_transform(X)
df = pd.DataFrame(X3, columns=["PC1","PC2","PC3"])
df["Species"] = iris.target_names[y]
sns.pairplot(df, hue="Species")
plt.show()
```

```python
pca2 = PCA(n_components=2)
X2 = pca2.fit_transform(X)

df2 = pd.DataFrame(X2, columns=["PC1", "PC2"])
df2["Species"] = iris.target_names[y]

sns.pairplot(df2, hue="Species")
plt.show()
```

```python
print(pca2.explained_variance_ratio_)
print("Total variance:", pca2.explained_variance_ratio_.sum())
```

```python
plt.scatter(X2[:,0], X2[:,1], c=y)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Iris Dataset: PCA 4D → 2D")
plt.show()
```
